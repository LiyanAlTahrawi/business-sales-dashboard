```python
from google.colab import files
uploaded = files.upload()
```

```python
import pandas as pd

file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name, encoding='latin1')
df.head()
```

```python
print(df.shape)
print(df.columns)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.columns = df.columns.str.lower().str.strip().str.replace(" ", "_").str.replace("-", "_")
df.head()
```

```python
df.isnull().sum()
```

```python
print(df.duplicated().sum())
df = df.drop_duplicates()
print(df.duplicated().sum())
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])
df['ship_date'] = pd.to_datetime(df['ship_date'])
df.info()
```

```python
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['month_name'] = df['order_date'].dt.month_name()
df['quarter'] = df['order_date'].dt.quarter
df.head()
```

```python
df['profit_margin'] = df['profit'] / df['sales']
df['shipping_days'] = (df['ship_date'] - df['order_date']).dt.days
df['sales_per_quantity'] = df['sales'] / df['quantity']
df.head()
```

```python
import numpy as np

df = df.replace([np.inf, -np.inf], np.nan)
df.isnull().sum()
```

```python
df = df.dropna() #just in case
df.isnull().sum()
```

```python
print(df[df['sales'] < 0].shape)
print(df[df['quantity'] <= 0].shape)
print(df[df['shipping_days'] < 0].shape)
```

```python
text_columns = ['category', 'sub_category', 'region', 'segment', 'ship_mode']

for col in text_columns:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip()

df.head()
```

```python
df.to_csv("cleaned_superstore_sales.csv", index=False)
```

```python
files.download("cleaned_superstore_sales.csv")
```

```python
monthly_sales = df.groupby(['year', 'month'])['sales'].sum().reset_index()
monthly_sales.head()
```

```python
monthly_sales.to_csv("monthly_sales.csv", index=False)
files.download("monthly_sales.csv")
```
